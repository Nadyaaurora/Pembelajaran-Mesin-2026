#### **Kegunaan Library**

1. **PyPREP (`pyprep`)**
    
    PyPREP adalah library Python yang digunakan untuk melakukan preprocessing pada sinyal EEG. Library ini membantu membersihkan sinyal dari noise dan gangguan sebelum data digunakan untuk proses analisis.

2. **SciPy (`scipy`)**
    
    SciPy adalah library Python yang digunakan untuk komputasi ilmiah dan numerik. SciPy menyediakan berbagai fungsi untuk aljabar linear, optimasi, statistik, pemrosesan sinyal, dan pengolahan data.

3. **wandb atau Weights & Biases (`wandb`)**
    
    wandb adalah library yang digunakan untuk membantu proses pengembangan model pembelajaran mesin. Library ini dapat digunakan untuk mencatat, melihat, dan memantau parameter, hasil, serta performa model selama proses pengembangan.

4. **pyECG (`pyecg`)**
    
    pyECG adalah library Python yang digunakan untuk pengolahan dan analisis data ECG. Library ini membantu memproses data ECG sebelum dilakukan analisis lebih lanjut.
